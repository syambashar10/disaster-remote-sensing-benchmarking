# STURM-Flood Full Dataset Verification Summary

This notebook summarizes the full-dataset verification results generated by:

`python -m disasterbench.tools.full_verify_sturm_flood`

The purpose is to document verified dataset reality before implementing converters/exporters.

```python
import json
import pandas as pd
from pathlib import Path

summary_path = Path('../outputs/verification/sturm_flood_full_verification_summary.json')
anomalies_path = Path('../outputs/verification/sturm_flood_full_verification_anomalies.csv')

with summary_path.open('r', encoding='utf-8') as f:
    report = json.load(f)

anomalies = pd.read_csv(anomalies_path)

report['dataset_id'], report['verification_type'], report['total_anomaly_count']
```

```python
rows = []

for sensor, info in report['sensors'].items():
    row = {
        'sensor': sensor,
        **info['counts'],
        **info['pairing'],
        'mask_values_global': info['mask_values_global'],
        'anomaly_count': info['anomaly_count'],
    }
    rows.append(row)

pd.DataFrame(rows)
```

```python
for sensor, info in report['sensors'].items():
    print('\n' + '=' * 80)
    print(sensor)
    print('=' * 80)
    print('Image size counts:', info['raster_distributions']['image_size_counts'])
    print('Mask size counts:', info['raster_distributions']['mask_size_counts'])
    print('Image band counts:', info['raster_distributions']['image_band_count_counts'])
    print('Mask band counts:', info['raster_distributions']['mask_band_count_counts'])
    print('Image dtype counts:', info['raster_distributions']['image_dtype_counts'])
    print('Mask dtype counts:', info['raster_distributions']['mask_dtype_counts'])
    print('Image CRS counts:', info['raster_distributions']['image_crs_counts'])
    print('Mask CRS counts:', info['raster_distributions']['mask_crs_counts'])
    print('Image nodata counts:', info['raster_distributions']['image_nodata_counts'])
    print('Mask nodata counts:', info['raster_distributions']['mask_nodata_counts'])
```
